Check duplicate track IDs

```python
# Check duplicate track IDs
duplicate_track_ids = df['track_id'].duplicated().sum()

print("Duplicate Track IDs:", duplicate_track_ids)
```

```python
# Find track IDs that appear more than once
duplicate_tracks = df[df['track_id'].duplicated(keep=False)]

# Count how many genres are associated with each track ID
genre_counts = duplicate_tracks.groupby('track_id')['track_genre'].nunique()

print("Track IDs appearing more than once:",
      duplicate_tracks['track_id'].nunique())

print("\nTrack IDs associated with multiple genres:",
      (genre_counts > 1).sum())
```

```python
duplicate_tracks[
    ['track_id', 'track_name', 'artists', 'track_genre']
].head(15)
```

```python
# Check duplicate combinations of track_id and track_genre
duplicate_track_genre = df.duplicated(
    subset=['track_id', 'track_genre']
).sum()

print("Duplicate Track ID + Genre combinations:",
      duplicate_track_genre)
```

```python
# Display duplicate track ID and genre combinations
duplicate_records = df[
    df.duplicated(
        subset=['track_id', 'track_genre'],
        keep=False
    )
]

duplicate_records[
    ['track_id', 'track_name', 'artists', 'track_genre']
].sort_values('track_id').head(20)
```

```python
# Remove duplicate track ID and genre combinations
df.drop_duplicates(
    subset=['track_id', 'track_genre'],
    keep='first',
    inplace=True
)

# Reset index
df.reset_index(drop=True, inplace=True)
```

```python
print("Dataset Shape:", df.shape)

print("Remaining Duplicate Track-Genre Combinations:",
      df.duplicated(
          subset=['track_id', 'track_genre']
      ).sum())
```

```python
# Convert duration from milliseconds to minutes
df['duration_minutes'] = df['duration_ms'] / 60000

# Display the 10 longest tracks
df.nlargest(10, 'duration_minutes')[
    ['track_name', 'artists', 'duration_ms',
     'duration_minutes', 'track_genre']
]
```

Top 10 Most Popular Artists by Average Popularity

```python
# Calculate average popularity for every artist
all_artist_popularity = (
    df.groupby('artists')['popularity']
    .agg(['mean', 'count'])
)

# Keep artists with at least 20 track records
all_artist_popularity = all_artist_popularity[
    all_artist_popularity['count'] >= 20
]

# Sort by average popularity
top_popular_artists = (
    all_artist_popularity
    .sort_values(by='mean', ascending=False)
    .head(10)
)

print(top_popular_artists)
```

```python
plt.figure(figsize=(12, 6))

sns.barplot(
    x=top_popular_artists['mean'],
    y=top_popular_artists.index
)

plt.title('Top 10 Artists by Average Popularity')
plt.xlabel('Average Popularity')
plt.ylabel('Artists')

plt.show()
```

Top 10 Albums by Average Popularity

```python
# Calculate average popularity for each album
album_popularity = (
    df.groupby('album_name')['popularity']
    .agg(['mean', 'count'])
)

# Consider albums with at least 5 track records
album_popularity = album_popularity[
    album_popularity['count'] >= 5
]

# Find top 10 albums
top_10_albums = (
    album_popularity
    .sort_values(by='mean', ascending=False)
    .head(10)
)

print(top_10_albums)
```

```python
plt.figure(figsize=(12, 6))

sns.barplot(
    x=top_10_albums['mean'],
    y=top_10_albums.index
)

plt.title('Top 10 Albums by Average Popularity')
plt.xlabel('Average Popularity')
plt.ylabel('Album Name')

plt.show()
```

Popularity distribution by duration category

```python
df['duration_category'] = pd.cut(
    df['duration_minutes'],
    bins=[0, 3, 5, 10, float('inf')],
    labels=['Short', 'Medium', 'Long', 'Very Long']
)

print(df.groupby(
    'duration_category',
    observed=True
)['popularity'].mean())
```

```python
plt.figure(figsize=(8, 5))

sns.barplot(
    x=df['duration_category'],
    y=df['popularity']
)

plt.title('Average Track Popularity by Duration')
plt.xlabel('Duration Category')
plt.ylabel('Average Popularity')

plt.show()
```
